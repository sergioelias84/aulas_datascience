# if / else

if: Testa a condição e apresenta a condição.

else: Representa o caminho alternativo; O else, atende caso nenhuma condição anterior esteja verdadeira.

```python
nota = 6.5
if nota >= 7:
  print('Aprovado')
else:
  print('Reprovado')
```

# elif (senão)

Usa quando existem mais de duas possibilidades; O elif significa "senão, se", ou seja;

Ele oferece outra condição quando as anteriores não foram atendidas.



```python
nota = 7
if nota >= 7:
  print('Aprovado')
elif nota >= 5:
  print('Recuperação')
else:
  print('Reprovado')
```

# Base de alunos
Vamos classificar os alunos por meio da nossa base de dados aluno_desempenho.csv

```python
import pandas as pd
arquivo = '/content/alunos_desempenho.csv'
df = pd.read_csv(arquivo)
df.head()
```

# Criando uma função para classificar cada aluno em media e frenquncia

```python
def classificar_aluno(media, frequencia):
    if media >= 7 and frequencia >= 75:
        return 'Aprovado'
    elif media < 7 and frequencia >= 75:
        return 'Reprovado por média'
    elif media >= 7 and frequencia < 75:
        return 'Reprovado por frequência'
    else:
        return 'Reprovado por média e frequência'

df['media'] = (df['nota_1'] + df['nota_2']) / 2

df['situacao'] = df.apply(
    lambda row: classificar_aluno(row['media'], row['frequencia']),
    axis=1
)

df.head()
```

Codigo completo

```python
import pandas as pd

df = pd.read_csv("pedidos_loja.csv")

def calcular_desconto(valor, tipo_cliente, usou_cupom):
    if tipo_cliente == "vip":
        return 0.15
    elif valor > 150:
        return 0.10
    elif usou_cupom:
        return 0.05
    else:
        return 0

df["desconto"] = df.apply(
    lambda linha: calcular_desconto(
        linha["valor_compra"],
        linha["tipo_cliente"],
        linha["usou_cupom"]
    ),
    axis=1
)

df["valor_desconto"] = (df["valor_compra"] * df["desconto"]).round(2)

df["valor_final"] = (df["valor_compra"] - df["valor_desconto"]).round(2)

print(df)
```
