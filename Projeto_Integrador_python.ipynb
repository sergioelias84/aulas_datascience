```python
import pandas as pd

df = pd.read_csv("vendas_semana.csv")
```

# 1. Total de vendas



```python
total_vendas = 0

for venda in df["vendas"]:
    total_vendas += venda
    print(f"Total de vendas: R$ {total_vendas:.2f}")
```

# 2. Média diária


```python
media_diaria = df["vendas"].mean()
print(f"Média diária: R$ {media_diaria:.2f}")
```

# 3. Maior venda


```python
maior_venda = df["vendas"].max()
dia_maior_venda = df.loc[df["vendas"].idxmax(), "dia"]
print(f"Maior venda: R$ {maior_venda:.2f}")
```

# 4. Dias que bateram a meta


```python
dias_meta = df[df["vendas"] >= df["meta"]]
print("\nDias que bateram a meta:")
for dia in dias_meta["dia"]:
    print(f"- {dia}")
```

# Relatório de vendas;

```python
print("RELATÓRIO DE VENDAS")
print("-" * 30)
print(f"Total de vendas: R$ {total_vendas:.2f}")
print(f"Média diária: R$ {media_diaria:.2f}")
print(f"Maior venda: R$ {maior_venda:.2f}")
print(f"Dia da maior venda: {dia_maior_venda}")
print("\nDias que bateram a meta:")

for dia in dias_meta["dia"]:
    print(f"- {dia}")
```
